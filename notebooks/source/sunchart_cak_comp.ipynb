```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import Normalize
import matplotlib as mpl
mpl.style.use('bkj_style.mplstyle')

# Initialize matrices (101 years from 1907-2007, 366 days)
bt1 = np.zeros((101, 366), dtype=int)  # Dataset 1
bt2 = np.zeros((101, 366), dtype=int)  # Dataset 2
bt_combined = np.zeros((101, 366), dtype=int)  # Combined (0=no data, 1=DS1, 2=DS2, 3=both)

def mark_date1(date_str):
    try:
        d = datetime.strptime(str(date_str), "%Y-%m-%d")
        year_idx = d.year - 1907
        day_idx = (d - datetime(d.year, 1, 1)).days
        if 0 <= year_idx < 101 and 0 <= day_idx < 366:
            bt1[year_idx, day_idx] = 1
            bt_combined[year_idx, day_idx] += 1  # DS1 = +1
    except ValueError:
        pass

def mark_date2(date_str):
    try:
        d = datetime.strptime(date_str, "%Y-%m-%dT%H:%M:%S.%f")
        year_idx = d.year - 1907
        day_idx = (d - datetime(d.year, 1, 1)).days
        if 0 <= year_idx < 101 and 0 <= day_idx < 366:
            bt2[year_idx, day_idx] = 1
            bt_combined[year_idx, day_idx] += 2  # DS2 = +2
    except ValueError:
        pass

# --- Load and process data ---
# Dataset 1
sunchart_file = 'dates_sunchart.txt'
sts_sunchart = pd.read_csv(sunchart_file, delim_whitespace=True, index_col=0)
for date_str in sts_sunchart.index:
    mark_date1(str(date_str))

# Dataset 2
data = pd.read_csv('data/corrected_time_kodai_v6.txt', 
                  skiprows=10, delim_whitespace=True, header=None, usecols=[1])
for date_str in data[1]:
    mark_date2(date_str)

# --- Ensure bt_combined is 0, 1, 2, or 3 ---
bt_combined = np.clip(bt_combined, 0, 3)

# --- Plotting ---
plt.figure(figsize=(8, 3))

# Use Viridis colormap (scaled 0-3)
norm = Normalize(vmin=0, vmax=3)
cmap = plt.get_cmap("inferno")#inferno,magma

# Plot each day as a colored square
for year_idx in range(101):
    year = 1907 + year_idx
    for day_idx in range(366):
        val = bt_combined[year_idx, day_idx]
        if val > 0:  
            plt.fill_between(
                [day_idx - 0.5, day_idx + 0.5],
                [year - 0.5, year - 0.5],
                [year + 0.5, year + 0.5],
                color=cmap(norm(val)),
                edgecolor="none"
            )

# --- Add color bar ---
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ticks=[0, 1, 2, 3])
cbar.set_ticklabels(["No Data", "Sunchart", "Ca II K", "Both"])
cbar.ax.tick_params(labelsize=9)  


plt.yticks(np.arange(1907, 2008, 15))
plt.xticks(
    np.arange(0, 366, 30.5),
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
)
plt.xlabel('Day of Year', fontsize=10)
plt.ylabel('Year', fontsize=10)
plt.grid()
#plt.grid(axis='y', alpha=0.3, linestyle="--")
plt.ylim(1906.5, 2007.5)
plt.xlim(-1, 366)
plt.tick_params(axis='both', labelsize=9)
plt.tight_layout()
plt.savefig("suncharts_cak_month_var.pdf", format="pdf", dpi=300, bbox_inches='tight')
plt.show()
```
