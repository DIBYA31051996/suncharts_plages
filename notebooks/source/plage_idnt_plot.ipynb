```python
import h5py
from matplotlib import pyplot as plt
import matplotlib as mpl
mpl.style.use('bkj_style.mplstyle')

with h5py.File('/home/dibya/ML_project/iou_plage_idnt.h5', "r") as f:
        # Read and preprocess mask
    
        epochs = f["mask"].attrs["epochs"]
        loss = f["mask"].attrs["loss"]
        val_loss = f["mask"].attrs["val_loss"]
        acc = f["mask"].attrs["acc"]
        val_acc = f["mask"].attrs["val_acc"]
        
fig=plt.figure(figsize = (6,5))
ax1 = fig.add_axes([0.01,0.01,0.5, 0.5])
ax2 = fig.add_axes([0.52,0.01,0.5, 0.5])

ax1.plot(epochs, loss, 'y', label='Training loss',color='darkgoldenrod')
ax1.plot(epochs, val_loss, 'r', label='Validation loss',color='indianred')
ax1.set_xlabel('Epochs',fontsize=16)
ax1.set_ylabel('Loss',fontsize=16)
ax1.tick_params(axis='both',labelsize=14)
ax1.set_xticks([0,10,20,30,40,50,60])
ax1.set_yticks([0.25,0.5,0.75,1,1.25,1.5])
ax1.annotate('(a)', (2,1.35), fontsize=16)
ax1.legend(fontsize=12,loc=1,handlelength=1.1,frameon= False)

ax2.plot(epochs, acc, 'y', label='Training IoU',color='darkgoldenrod')
ax2.plot(epochs, val_acc, 'r', label='Validation IoU',color='indianred')
ax2.set_xlabel('Epochs',fontsize=16)
ax2.set_ylabel('IoU',fontsize=16)
ax2.set_ylim(0,1)
ax2.tick_params(axis='both',labelsize=14,labelleft=False,labelright=True)
ax2.yaxis.set_label_position("right")
ax2.set_xticks([0,10,20,30,40,50,60])
ax2.annotate('(b)', (1,0.9), fontsize=16)
ax2.legend(fontsize=12,loc=1,handlelength=1.1,frameon= False)

plt.savefig('iou_plage.pdf', bbox_inches='tight', pad_inches=0, dpi=300)
plt.show()
```
