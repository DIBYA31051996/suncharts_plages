# Solar Disk Circle Detection

Post-process predicted limb masks and estimate solar-disk circles.

**Configuration:** Mask and image paths are configuration values; update them for the local dataset.

```python
# Step 1: Generate or inspect model predictions.
mask_directory = '/home/dibya/limb_prediction1/'
```

```python
# Step 2: Run the next processing stage.
mask_dataset=[]
```

```python
# Step 3: Import dependencies and initialise this stage.
import os
import cv2
import numpy as np
from PIL import Image

png_files = [os.path.join(root, filename) for root, _, files in os.walk(mask_directory) for filename in files if filename.endswith('.png')]
png_files.sort()

for image_path in png_files:
            image = cv2.imread(image_path, cv2.COLOR_BGR2RGB)
            image = Image.fromarray(image)
            mask_dataset.append(np.array(image))
            
```

```python
# Step 4: Run the next processing stage.
name_image = []
for root, dirs, files in os.walk(mask_directory):
    for image_name in files:
        if image_name.lower().endswith('.png'):
            image_path = os.path.join(root, image_name)
            name_image.append(image_name[:-4])
            name_image.sort()
print(name_image)
```

```python
# Step 10: Import dependencies and initialise this stage.
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2 as cv2
import glob
import os, sys
import imutils
```

```python
# Step 11: Visualise intermediate or final results.
tilt_angle = []
image_no = []

for i in range(0,len(masks)):

    flipped_image = cv2.flip(masks[i], 0)

    # Apply Canny edge detection to detect edges
    edges = cv2.Canny(flipped_image, 0,3, apertureSize=3)
    kernel = np.ones((1,1), np.uint8)
    closing = cv2.morphologyEx(edges, cv2.MORPH_CLOSE, kernel, iterations=2)
    erosion = cv2.erode(closing, kernel, iterations=2)
    bg = cv2.dilate(erosion, kernel, iterations = 2)

    # # Find lines using Hough Transform
    lines = cv2.HoughLines(flipped_image, 1, np.pi/180, 10) 
    
    # Extract the first detected line
    rho, theta = lines[0][0]
    
    # Calculate the tilt of the line
    tilt_radians = np.pi/2 - theta  # Angle with the x-axis in radians
    tilt_degrees = np.degrees(tilt_radians)  # Convert radians to degrees

    
    
   #Calculate tilt of the lines
    tilt_angle.append(tilt_degrees)
    image_no.append(i)
    
    print(tilt_degrees)
```

```python
# Step 15: Import dependencies and initialise this stage.
from skimage import io, color, measure, draw, img_as_bool
import numpy as np
from scipy import optimize
import matplotlib.pyplot as plt
from skimage.transform import hough_circle, hough_circle_peaks

x0 = []
y0 = []
rad = []

x0_main = []
y0_main = []
rad_main = []

for i in range(0,len(masks)):

    flipped_image_ = cv2.flip(masks[i], 0)
    if tilt_angle[i] > 0:
         Rotated1_image = imutils.rotate(masks[i], angle=-90.0+tilt_angle[i])
    if tilt_angle[i]  < 0:
         Rotated1_image = imutils.rotate(masks[i], angle=-90.0-tilt_angle[i])

    # Apply Canny edge detection to detect edges
    med = cv2.medianBlur(Rotated1_image,1)
    edges_ = cv2.Canny(med, 0,3,apertureSize=3)
   
    kernel_ = np.ones((1,1), np.uint8)
    closing_ = cv2.morphologyEx(edges_, cv2.MORPH_CLOSE, kernel_, iterations=2)
    erosion_ = cv2.erode(closing_, kernel_, iterations=2)
    bg_ = cv2.dilate(erosion_, kernel_, iterations = 2)

    detected_circles=cv2.HoughCircles(bg_, cv2.HOUGH_GRADIENT,1.0,300,param1=10,param2=20,minRadius=70,maxRadius=130)
    
    # Detect two radii
    hough_radii = np.arange(100, 103, 1)#100,101
    hough_res = hough_circle(edges_, hough_radii)
    
    # Select the most prominent 3 circles
    accums, a, b, r = hough_circle_peaks(hough_res, hough_radii, total_num_peaks=3)
    
    
    x0.append(int(a[0]))
    y0.append(int(b[0]))
    rad.append(int(r[0]))

    x0_main.append(np.round((a[0]/256.0)*6144.0))
    y0_main.append(np.round((b[0]/256.0)*6144.0))
    rad_main.append(np.round((r[0]/256.0)*6144.0))

    print(int(a[0]),int(b[0]),int(r[0]))
```

```python
# Step 16: Run the next processing stage.
with open ("1909.txt",'w') as file:
    for file_name,center_x,center_y,radius,p_angle in zip (name_image,x0_main,y0_main,rad_main,np.round(tilt_angle,5)): 
        line = f"{file_name}      {center_x}    {center_y}    {radius}     {p_angle}\n"
        file.write(line)
```
